# Radar Simulation Demo

This notebook demonstrates synthetic range-Doppler heatmap generation and visualization using `radar_simulation.py`.

Run the cells below to generate sample heatmaps for different scenarios.

```python
import os
from radar_simulation import generate_range_doppler_heatmap, apply_fft, visualize_heatmap

os.makedirs('data/demo', exist_ok=True)
# Metal object scenario
hm_metal = generate_range_doppler_heatmap(128, 64, scenario='metal_object', metal=True, clutter_level=0.05)
visualize_heatmap(hm_metal, title='Metal object (range-Doppler)')

# Empty room + clutter scenario
hm_clutter = generate_range_doppler_heatmap(128, 64, scenario='clutter', metal=False, clutter_level=0.2)
visualize_heatmap(hm_clutter, title='Clutter scene')

# FFT transform preview
visualize_heatmap(apply_fft(hm_metal), title='FFT (log-magnitude) of metal scene')
```
